```python
import numpy as np
import tensorflow as tf

def a_regular_function(x, y, b):
  x = tf.matmul(x, y)
  x = x + b
  return x

a_function_that_user_a_graph = tf.function(a_regular_function)

x1 = tf.constant([[1.0, 2.0]])
y1 = tf.constant([[2.0], [3.0]])
b1 = tf.constant(4.0)

orig_value = a_regular_function(x1, y1, b1).numpy()
print(orig_value)

tf_function_value = a_function_that_user_a_graph(x1, y1, b1).numpy()
print(orig_value)
print(tf_function_value)
```

```python
import numpy as np
import tensorflow as tf

class SimpleModule(tf.Module):
  def __init__(self, name=None):
    super() .__init__(name=name)
    self.a_variable = tf.Variable(5.0, name="train_me")
    self.non_trainable_variable = tf.Variable(5.0, trainable=False, name="do_not_train")
  def __call__(self, x):
    return self.a_variable * x + self.non_trainable_variable

simple_module = SimpleModule(name = "simple")

rs = simple_module(tf.constant(5.0))
print(rs)
print('Train variable', simple_module.trainable_variables)
print('All variable', simple_module.variables)
```

```python
import numpy as np
import tensorflow as tf

chkp_path = "my_checkpoint"
checkpoint = tf.train.Checkpoint(model=simple_module)
checkpoint.write(chkp_path)
```

```python
import numpy as np
import tensorflow as tf

class SequentialModule(tf.Module):
    def __init__(self, name=None):
        super().__init__(name=name)
        self.w = tf.Variable([[0.36512682, 0.0],
                              [0.36512682, 0.0],
                              [0.36512682, 0.0]], name='w')
        self.b = tf.Variable([0.0, 0.0], name='b')

    def __call__(self, x):
        return tf.matmul(x, self.w) + self.b

new_model = SequentialModule()
new_checkpoint = tf.train.Checkpoint(model=new_model)
new_checkpoint.restore("my_checkpoint").expect_partial()

print(new_model(tf.constant([[2.0, 2.0, 2.0]])))
```

```python
import tensorflow as tf

print("--- Bài 12 ---")

x = tf.random.uniform([3, 2, 2], minval=1, maxval=10)
y = tf.random.uniform([3, 2, 2], minval=1, maxval=10)

@tf.function
def tinh_toan_graph(a, b):
    tong = a + b
    hieu = a - b
    tich = a * b
    thuong = a / b
    return tong, hieu, tich, thuong

kq_tong, kq_hieu, kq_tich, kq_thuong = tinh_toan_graph(x, y)
print("Tổng:\n", kq_tong.numpy())
print("Hiệu:\n", kq_hieu.numpy())

print("\n--- Bài 13 ---")

class ModelTong(tf.Module):
    @tf.function
    def __call__(self, a, b):
        return a + b

model_tong = ModelTong()
print("Tổng x + y:\n", model_tong(x, y).numpy())

print("\n--- Bài 14 ---")

class ModelHieuTich(tf.Module):
    @tf.function
    def __call__(self, a, b):
        hieu = a - b
        tich = a * b
        return hieu, tich

model_hieu_tich = ModelHieuTich()
kq_h, kq_t = model_hieu_tich(x, y)
print("Hiệu x - y:\n", kq_h.numpy())
print("Tích x * y:\n", kq_t.numpy())


print("\n--- Bài 15 ---")

class ModelChuVi(tf.Module):
    @tf.function
    def __call__(self, a, b):
        return (a + b) * 2

model_cv = ModelChuVi()
chieu_dai = tf.constant([5.0, 10.0])
chieu_rong = tf.constant([3.0, 4.0])
print("Chu vi P:", model_cv(chieu_dai, chieu_rong).numpy())


print("\n--- Bài 16 ---")

class ModelDienTichHộp(tf.Module):
    @tf.function
    def __call__(self, a, b, c):
        s_xung_quanh = (a + b) * c * 2
        s_toan_phan = (a * b) * 2 + s_xung_quanh
        return s_xung_quanh, s_toan_phan

model_hop = ModelDienTichHộp()
a = tf.constant([2.0])
b = tf.constant([3.0])
c = tf.constant([4.0])

s_xq, s_tp = model_hop(a, b, c)
print("Diện tích xung quanh:", s_xq.numpy())
print("Diện tích toàn phần:", s_tp.numpy())
```

```python
import tensorflow as tf

print("--- Bài 17 ---")

class ModelChungMinhSongSong(tf.Module):
    @tf.function
    def __call__(self, AD, BC, AB, CD):
        dieu_kien_1 = tf.equal(AD, BC)
        dieu_kien_2 = tf.equal(AB, CD)

        ad_song_song_bc = tf.logical_and(dieu_kien_1, dieu_kien_2)
        return ad_song_song_bc

model_17 = ModelChungMinhSongSong()

AD_test = tf.constant([5.0, 5.0])
BC_test = tf.constant([5.0, 4.0])
AB_test = tf.constant([3.0, 3.0])
CD_test = tf.constant([3.0, 3.0])

print("Kết quả chứng minh AD // BC:", model_17(AD_test, BC_test, AB_test, CD_test).numpy())


print("\n--- Bài 18 ---")

class ModelChungMinhBangNhau(tf.Module):
    @tf.function
    def __call__(self, AD, BC, ad_song_song_bc):
        dieu_kien_1 = tf.equal(AD, BC)
        dieu_kien_2 = ad_song_song_bc

        ab_bang_cd = tf.logical_and(dieu_kien_1, dieu_kien_2)
        return ab_bang_cd

model_18 = ModelChungMinhBangNhau()

AD_test2 = tf.constant([6.0, 6.0])
BC_test2 = tf.constant([6.0, 6.0])
song_song_test = tf.constant([True, False])

print("Kết quả chứng minh AB = CD:", model_18(AD_test2, BC_test2, song_song_test).numpy())
```

```python
import tensorflow as tf

print("--- Bài 19 ---")

class ModelChungMinhTrungDiem(tf.Module):
    @tf.function
    def __call__(self, AM, MB, AN, NC, mn_song_song_bc):
        dieu_kien_1 = tf.equal(AM, MB)

        dieu_kien_2 = mn_song_song_bc

        la_trung_diem = tf.logical_and(dieu_kien_1, dieu_kien_2)

        return la_trung_diem

model_19 = ModelChungMinhTrungDiem()

AM_test = tf.constant([4.0, 4.0])
MB_test = tf.constant([4.0, 5.0])
AN_test = tf.constant([3.0, 3.0])
NC_test = tf.constant([3.0, 3.0])
song_song_test = tf.constant([True, True])

print("Kết quả chứng minh NA = NC:", model_19(AM_test, MB_test, AN_test, NC_test, song_song_test).numpy())
```
